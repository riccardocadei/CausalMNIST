```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import generalization results 
pW=0.5
pU=0.02
e=1
exp = "RCT"
estimator = "CF"
results = pd.read_csv(f'results/{e}/{pW}/{pU}/{exp}/gen_{estimator}.csv', index_col=0)
# filter high accuracy
# results = results[results['acc'] > 0.60]
results['bias'] = 0 - results['PPAIPW_OC']
results_grouped = results.drop(['seed'], axis=1).groupby(['method', 'target']).agg(['mean', 'std'])
results_grouped = results_grouped.reset_index()
results_grouped
results_grouped.loc[results_grouped['target'] == 'A', ('bias', 'mean')] += 1.5
results_grouped.loc[results_grouped['target'] == 'D', ('bias', 'mean')] += 1.2
results_grouped.loc[results_grouped['target'] == 'E', ('bias', 'mean')] += 0.75
results_grouped = results_grouped.set_index(['method', 'target'])

results_grouped['bias']
```

```python
# print scatterplot Bias vs Acc coloring by method and shaping by target
plt.figure(figsize=(8, 6))
sns.scatterplot(x='bias', y='acc', hue='method', data=results, style='target')
plt.xlabel('Bias')
plt.ylabel('Accuracy');
```
