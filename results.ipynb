```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pW=0.5
pU=0.02
e=3
exp = "RCT"
df = pd.read_csv(f'results/{e}/{pW}/{pU}/{exp}/generalization.csv', index_col=0)
df['target'] = df['target'].astype('category')

# group by target and get mean and std ATE
df_grouped = df.groupby('target').agg({'ATE': ['mean', 'std']}).reset_index()
df_grouped.columns = ['target', 'mean_ate', 'std_ate']
df_grouped['target'] = df_grouped['target'].astype(str)
df_grouped['target'] = df_grouped['target'].str.replace('target_', '')
df_grouped['target'] = df_grouped['target'].str.replace('_', ' ')
df_grouped
```

```python
# import generalization results 
pW=0.5
pU=0.02
e=3
exp = "RCT"
results = pd.read_csv(f'results/{e}/{pW}/{pU}/{exp}/generalization.csv', index_col=0)
results['bias'] = 0.75 - results['PPAIPW_OC']
results_grouped = results.drop(['seed'], axis=1).groupby(['method', 'target']).agg(['mean', 'std'])
results_grouped['bias']
```

```python
# print scatterplot Bias vs Acc coloring by method and shaping by target
plt.figure(figsize=(8, 6))
sns.scatterplot(x='bias', y='acc', hue='method', data=results, style='target')
plt.xlabel('Bias')
plt.ylabel('Accuracy');
```
